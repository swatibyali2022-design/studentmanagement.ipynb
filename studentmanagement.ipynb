```python
# Student Management Application

students = []
next_id = 1


# Function to add a student
def add_student(name, email, department, year):
    global next_id

    # Validate name
    if not name or not name.strip():
        return "Error: Name is required."

    # Validate email
    if not email or not email.strip():
        return "Error: Email is required."

    # Basic email validation
    if "@" not in email or "." not in email:
        return "Error: Please enter a valid email."

    # Validate year
    if not str(year).isdigit():
        return "Error: Year must be a number."

    year = int(year)

    if year < 1 or year > 5:
        return "Error: Year must be between 1 and 5."

    # Check duplicate email
    for student in students:
        if student["email"].lower() == email.strip().lower():
            return "Error: A student with this email already exists."

    # Create student
    student = {
        "id": next_id,
        "name": name.strip(),
        "email": email.strip(),
        "department": department.strip(),
        "year": year
    }

    students.append(student)
    next_id += 1

    return f"Student added successfully. Student ID: {student['id']}"


# Function to view all students
def view_students():
    if not students:
        print("\nNo students found.")
        return

    print("\n========== STUDENT LIST ==========")

    for student in students:
        print(f"ID         : {student['id']}")
        print(f"Name       : {student['name']}")
        print(f"Email      : {student['email']}")
        print(f"Department : {student['department']}")
        print(f"Year       : {student['year']}")
        print("----------------------------------")


# Function to update a student by ID
def update_student(student_id, name=None, email=None, department=None, year=None):

    # Find student
    for student in students:

        if student["id"] == student_id:

            # Validate name if provided
            if name is not None:
                if not name.strip():
                    return "Error: Name cannot be empty."
                student["name"] = name.strip()

            # Validate email if provided
            if email is not None:
                if not email.strip():
                    return "Error: Email cannot be empty."

                if "@" not in email or "." not in email:
                    return "Error: Please enter a valid email."

                # Check if email belongs to another student
                for other_student in students:
                    if (other_student["id"] != student_id and
                            other_student["email"].lower() == email.strip().lower()):
                        return "Error: This email is already used by another student."

                student["email"] = email.strip()

            # Update department
            if department is not None:
                if not department.strip():
                    return "Error: Department cannot be empty."
                student["department"] = department.strip()

            # Update year
            if year is not None:
                if not str(year).isdigit():
                    return "Error: Year must be a number."

                year = int(year)

                if year < 1 or year > 5:
                    return "Error: Year must be between 1 and 5."

                student["year"] = year

            return "Student updated successfully."

    return "Error: Student ID not found."


# Function to delete a student
def delete_student(student_id):

    for student in students:

        if student["id"] == student_id:
            students.remove(student)
            return "Student deleted successfully."

    return "Error: Student ID not found."


# Main application
def main():

    while True:

        print("\n========== STUDENT MANAGEMENT SYSTEM ==========")
        print("1. Add Student")
        print("2. View Students")
        print("3. Update Student")
        print("4. Delete Student")
        print("5. Exit")

        choice = input("Enter your choice: ")

        # Add student
        if choice == "1":

            print("\n--- Add Student ---")

            name = input("Enter name: ")
            email = input("Enter email: ")
            department = input("Enter department: ")
            year = input("Enter year: ")

            result = add_student(
                name,
                email,
                department,
                year
            )

            print(result)

        # View students
        elif choice == "2":
            view_students()

        # Update student
        elif choice == "3":

            print("\n--- Update Student ---")

            try:
                student_id = int(input("Enter student ID: "))
            except ValueError:
                print("Error: Student ID must be a number.")
                continue

            print("Leave a field empty if you don't want to change it.")

            name = input("Enter new name: ")
            email = input("Enter new email: ")
            department = input("Enter new department: ")
            year = input("Enter new year: ")

            # Convert empty inputs to None
            name = name if name.strip() else None
            email = email if email.strip() else None
            department = department if department.strip() else None
            year = year if year.strip() else None

            result = update_student(
                student_id,
                name,
                email,
                department,
                year
            )

            print(result)

        # Delete student
        elif choice == "4":

            print("\n--- Delete Student ---")

            try:
                student_id = int(input("Enter student ID: "))
            except ValueError:
                print("Error: Student ID must be a number.")
                continue

            result = delete_student(student_id)

            print(result)

        # Exit
        elif choice == "5":

            print("\nThank you for using Student Management System!")
            break

        else:
            print("Error: Invalid choice. Please select 1-5.")


# Start the application
if __name__ == "__main__":
    main()
```
